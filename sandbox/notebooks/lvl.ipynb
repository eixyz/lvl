### Estimate refractor depth

```python
# Data
import numpy as np
v0 = 536
v1 = 1840
ti1 = 21.55
v2 = 3268
ti2=41.37

def compute_depths(v0, v1, ti1, v2=None, ti2=None):
    # 2 layer depth
    h0 = (ti1*v0*v1)/(2000*np.sqrt(v1**2 - v0**2))

    # 3 layer depth
    if v2 is not None and ti2 is not None:
        ti2_eff = ti2 - (2000*h0*np.cos(np.arcsin(v0/v1)))/v0
        h1 = (v1*ti2_eff)/(2000*np.cos(np.arcsin(v1/v2)))
        refractor_depth = h0 + h1
        return h0, h1, refractor_depth
    return h0, None, None

h0, h1, refractor_depth = compute_depths(v0, v1, ti1, v2, ti2)
print(f"Computed depths: h0 = {h0:.2f} m, h1 = {h1 if h1 is not None else 'N/A':.2f} m, refractor depth = {refractor_depth if refractor_depth is not None else 'N/A':.2f} m")
```
